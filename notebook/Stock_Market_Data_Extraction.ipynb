```python
import yfinance as yf
import pandas as pd

tickers = [
    "NFLX", "DIS", "LYV", "WBD", "FOXA", "FWONA", "WMG", "CMCSA", "TKO", "ROKU", "PARA", "LGF-B", "SONY", "IMAX", "CNK"

]

first_wave_start = "2020-01-01"
first_wave_end = "2020-06-30"

second_wave_start = "2020-07-01"
second_wave_end = "2020-12-31"

first_wave_data = yf.download(tickers, start=first_wave_start, end=first_wave_end, group_by='ticker')

second_wave_data = yf.download(tickers, start=second_wave_start, end=second_wave_end, group_by='ticker')

# Combine both waves of data into one list
all_data = []

for ticker in tickers:
    df = first_wave_data[ticker].copy()
    df["Ticker"] = ticker  # Add ticker column
    df["Wave"] = "First Wave"  # Label the first wave data
    all_data.append(df)

for ticker in tickers:
    df = second_wave_data[ticker].copy()
    df["Ticker"] = ticker  # Add ticker column
    df["Wave"] = "Second Wave"  # Label the second wave data
    all_data.append(df)

final_df = pd.concat(all_data)

final_df.reset_index(inplace=True)

final_df.to_csv("Stock_Market_Data_.csv", index=False)

print("✅ Stock data successfully saved to 'Stock_Market_Data.csv'")


```
